### Contexto biológico
Um gene (DNA) é **transcrito** em mRNA e depois **traduzido** em uma proteína. Na tradução, o mRNA é lido em grupos de **3 bases** chamados **códons**, e cada códon corresponde a um aminoácido. **Gabarito — Exercício 5, Parte A**

```python
class Gene:
    def __init__(self, sequencia):
        self.sequencia = sequencia          # passa pelo setter (validação)

    @property
    def sequencia(self):
        return self._sequencia

    @sequencia.setter
    def sequencia(self, valor):
        valor = valor.upper()
        if not all(base in "ATCG" for base in valor):
            raise ValueError("A sequência de DNA só pode conter A, T, C e G.")
        self._sequencia = valor

    def __len__(self):
        return len(self.sequencia)

    def conteudo_gc(self):
        gc = self.sequencia.count("G") + self.sequencia.count("C")
        return gc / len(self.sequencia) * 100

    def transcrever(self):
        return self.sequencia.replace("T", "U")

    def traduzir(self):
        mrna = self.transcrever()
        proteina = []
        for i in range(0, len(mrna) - 2, 3):     # passos de 3; ignora sobra incompleta
            codon = mrna[i:i + 3]
            aminoacido = CODIGO_GENETICO[codon]
            if aminoacido == "*":                # códon de parada
                break
            proteina.append(aminoacido)
        return "".join(proteina)

    def __repr__(self):
        return f"Gene({self.sequencia!r})"


# Teste rápido:
g = Gene("ATGGCCTTTTAA")
print("Gene:", g)
print("mRNA:", g.transcrever())
print("Proteína:", g.traduzir())
print(f"GC: {g.conteudo_gc():.1f}%")
```

**Gabarito — Exercício 5, Parte B**

```python
from functools import reduce

genes = [
    Gene("ATGGCCTTTTAA"),
    Gene("ATGAAACCCGGGTAA"),
    Gene("ATGCGCGCGCGCTAA"),
]

# (a) filter
ricos_em_gc = list(filter(lambda g: g.conteudo_gc() >= 50, genes))

# (b) map
proteinas = list(map(lambda g: g.traduzir(), genes))

# (c) reduce
comprimento_total = reduce(lambda total, g: total + len(g), genes, 0)

print("Genes ricos em GC (>=50%):", ricos_em_gc)   # [Gene('ATGCGCGCGCGCTAA')]
print("Proteínas:", proteinas)                     # ['MAF', 'MKPG', 'MRAR']
print("Total de bases:", comprimento_total)        # 42
```
