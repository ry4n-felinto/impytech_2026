## Exercício 1 🟢 — Regressão linear simples e avaliação de métricas

O dataset `load_linnerud` (já incluso no `scikit-learn`) traz medidas de 20 homens de um clube de ginástica: variáveis de **exercício** (`Chins`, `Situps`, `Jumps`) e variáveis **fisiológicas** (`Weight`, `Waist`, `Pulse`).

Vamos usar o número de abdominais (`Situps`) para prever o **pulso em repouso** (`Pulse`) — quanto mais condicionamento físico, menor tende a ser o pulso de repouso.


```python
from sklearn.datasets import load_linnerud
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

dados = load_linnerud(as_frame=True)
df_linnerud = dados.data.join(dados.target)

df_linnerud[["Situps", "Pulse"]]
```

> 💡 **Recapitulando:**
> - $MSE = \frac{1}{n}\sum (y_i - \hat{y}_i)^2$ — quanto **menor**, melhor (0 = previsão perfeita).
> - $R^2$ — vai de 0 (modelo tão bom quanto prever sempre a média) a 1 (previsão perfeita); pode até ficar negativo se o modelo for pior que a média.

### Solução — Exercício 1

```python
from sklearn.datasets import load_linnerud
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

dados = load_linnerud(as_frame=True)
df_linnerud = dados.data.join(dados.target)

X_lin = df_linnerud[["Situps"]]
y_lin = df_linnerud["Pulse"]

X_tr_lin, X_te_lin, y_tr_lin, y_te_lin = train_test_split(X_lin, y_lin, test_size=0.3, random_state=42)

modelo_lin = LinearRegression()
modelo_lin.fit(X_tr_lin, y_tr_lin)

mse_treino_lin = mean_squared_error(y_tr_lin, modelo_lin.predict(X_tr_lin))
mse_teste_lin  = mean_squared_error(y_te_lin, modelo_lin.predict(X_te_lin))
r2_treino_lin  = r2_score(y_tr_lin, modelo_lin.predict(X_tr_lin))
r2_teste_lin   = r2_score(y_te_lin, modelo_lin.predict(X_te_lin))

print(f"MSE treino: {mse_treino_lin:.2f}   |   MSE teste: {mse_teste_lin:.2f}")
print(f"R²  treino: {r2_treino_lin:.3f}   |   R²  teste: {r2_teste_lin:.3f}")
# Com um dataset tão pequeno (20 pacientes), os números podem oscilar bastante
# dependendo da divisão treino/teste. O R² costuma ser baixo aqui, sugerindo que
# 'Situps' sozinho explica pouco da variação do pulso -- a relação é fraca.
```
