# Parte 4 — Programação Funcional

Na Parte 1 aprendemos a **organizar** código com objetos. Agora vamos aprender um estilo diferente e complementar, muito útil para **processar conjuntos de dados** 

A ideia central da programação funcional é tratar **funções como dados**: podemos guardá-las em variáveis, passá-las como argumento e recebê-las como resultado. Isso permite escrever transformações de dados de forma concisa e legível.


## 4.1 Iteráveis e funções de ordem superior

Um **iterável** é qualquer coisa por onde dá para percorrer item a item: uma lista, uma string, um `range`, e, como vimos, até nossa `SequenciaDNA`.

Uma **função de ordem superior** é uma função que **recebe outra função** como argumento (ou devolve uma função). Isso só é possível porque, em Python, funções são objetos como qualquer outro.


````python
# Exemplo clínico: converter uma lista de temperaturas de °C para °F.
def celsius_para_fahrenheit(c):
    return c * 9/5 + 32

conversor = celsius_para_fahrenheit    # sem os parênteses: guardamos a função, não o resultado
print("37°C em °F:", conversor(37))

# Uma função de ordem superior: recebe uma função e a aplica a cada item.
def aplicar_a_todos(funcao, valores):
    resultado = []
    for v in valores:
        resultado.append(funcao(v))
    return resultado

temperaturas_c = [36.5, 37.0, 38.2, 39.1]
print("Em °F:", aplicar_a_todos(celsius_para_fahrenheit, temperaturas_c))
````

### Funções anônimas: `lambda`

Às vezes precisamos de uma função pequena e descartável, só para usar na hora. Para isso existe a `lambda`: uma função de uma linha, sem nome.

```python
lambda parametro: expressao
```

É equivalente a definir uma função com `def`, mas mais curta. Compare:


````python
# Estas duas formas fazem exatamente a mesma coisa:
def dobro(x):
    return x * 2

dobro_lambda = lambda x: x * 2

print(dobro(21), dobro_lambda(21))

# lambda é ideal para usar "na hora", dentro de outra função:
pesos = [72.0, 88.5, 60.2]
pesos_em_libras = aplicar_a_todos(lambda kg: kg * 2.205, pesos)
print("Pesos em libras:", [round(p, 1) for p in pesos_em_libras])
````

## 4.2 `map` — transformar cada item

A função embutida `map` aplica uma função a **cada elemento** de um iterável, produzindo um novo conjunto de valores. É a versão pronta e otimizada do nosso `aplicar_a_todos`.

```python
map(funcao, iteravel)
```

`map` devolve um objeto "preguiçoso" (calcula sob demanda), então normalmente o envolvemos em `list(...)` para ver os resultados.


````python
temperaturas_c = [36.5, 37.0, 38.2, 39.1]

temperaturas_f = list(map(lambda c: round(c * 9/5 + 32, 1), temperaturas_c))
print("°C:", temperaturas_c)
print("°F:", temperaturas_f)

# map também funciona com métodos dos nossos objetos!
# Vamos reaproveitar a classe Paciente com IMC:
class Paciente:
    def __init__(self, nome, peso, altura):
        self.nome = nome
        self.peso = peso
        self.altura = altura
    def calcular_imc(self):
        return self.peso / (self.altura ** 2)

turma = [
    Paciente("Ana",   72.0, 1.65),
    Paciente("Bruno", 88.5, 1.80),
    Paciente("Cecilia", 55.0, 1.60),
]

# Um IMC para cada paciente, em uma linha:
imcs = list(map(lambda p: round(p.calcular_imc(), 1), turma))
print("IMCs da turma:", imcs)
````

Repare como `map` substitui um laço `for` inteiro por uma única linha expressiva: *"para cada paciente, calcule o IMC"*.


## 4.3 `filter` — selecionar itens

A função `filter` mantém apenas os elementos para os quais uma função devolve `True`. É a ferramenta natural para **triagem**: separar de uma lista só os casos que atendem a um critério.

```python
filter(funcao_que_devolve_True_ou_False, iteravel)
```


````python
# Triagem de febre: manter apenas as temperaturas >= 37.8 °C
temperaturas = [36.5, 37.0, 38.2, 39.1, 36.8, 38.5]

febris = list(filter(lambda t: t >= 37.8, temperaturas))
print("Temperaturas com febre:", febris)

# Triagem de pacientes com sobrepeso/obesidade (IMC >= 25):
com_imc_alto = list(filter(lambda p: p.calcular_imc() >= 25, turma))
print("Pacientes com IMC >= 25:", [p.nome for p in com_imc_alto])
````

**a diferença essencial entre `map` vs `filter`:**

- `map` **transforma** cada item (a quantidade de itens não muda).
- `filter` **seleciona** itens (o conteúdo não muda, mas alguns itens ficam de fora).

Elas se combinam muito bem: primeiro `filter` para selecionar, depois `map` para transformar o que sobrou.

## 4.4 `reduce` — combinar tudo em um só valor

Enquanto `map` e `filter` devolvem uma coleção, `reduce` **condensa** um iterável inteiro em **um único valor**, aplicando uma função acumuladora de par em par. É o que usamos para somar, multiplicar, encontrar o máximo, etc.

Ao contrário de `map` e `filter`, `reduce` **não** é embutida: precisamos importá-la do módulo `functools`.

```python
from functools import reduce
reduce(funcao_de_dois_argumentos, iteravel)
```

Como funciona a soma, passo a passo, para `[10, 20, 30, 40]`:
`((10 + 20) + 30) + 40` → `60 + 40` → `100`. O resultado de cada passo vira a entrada do próximo.


````python
from functools import reduce

# Dose total de um medicamento administrada ao longo do dia (mg):
doses = [500, 500, 750, 500]
dose_total = reduce(lambda acumulado, dose: acumulado + dose, doses)
print("Dose total no dia:", dose_total, "mg")

# Maior glicemia registrada (encontrando o máximo "na mão"):
glicemias = [88, 102, 140, 95, 130]
maior = reduce(lambda a, b: a if a > b else b, glicemias)
print("Maior glicemia:", maior, "mg/dL")

# IMC médio da turma, no estilo funcional:
imcs = list(map(lambda p: p.calcular_imc(), turma))
soma_imcs = reduce(lambda a, b: a + b, imcs)
media_imc = soma_imcs / len(imcs)
print(f"IMC médio da turma: {media_imc:.1f}")
````

> 💡 Na prática do dia a dia, para operações comuns o Python já oferece atalhos: `sum(doses)`, `max(glicemias)`, `min(...)`. O `reduce` é a ferramenta **geral** por trás dessa ideia — útil quando a forma de combinar os itens é específica do seu problema e não existe uma função pronta.
