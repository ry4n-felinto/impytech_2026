> ## **Problema 1 - Expressão Genética** 🟢

Suponha que você e seu grupo de pesquisa estão fazendo um levantamento para descobrir uma relação entre a expressividade genética e o tumor.
Neste levantamento, vocês escolheram dois genes específicos: `TP53` e `BRCA1`.



### 📝 **Problema 1.1**

**🎯 Objetivo:** Usando a biblioteca pandas leia o arquivo `dados_expressao_genes.csv`, visualize de forma simples a tabela.

```python
# TODO: leia o arquivo `dados_rastreamento_albert.csv`
df = pd.read_csv('dados_expressao_genes.csv')

# TODO: visualize de forma resumida a estrutura das tabelas
df.head()
```

### 📝 **Problema 1.2**

**🎯 Objetivo:** Usando uma filtragem simples, faça um histograma comparando a quantidade de amostras de cada grupo de controle para o genes `TP53`.

```python
#TODO: Filtre o dados 
geneTP53_df = df[df['Genes'] == 'TP53']

#TODO: Conte os dados de cada grupo 
geneTP53_counts = geneTP53_df.groupby('Grupo').size()

#TODO: Crie o gráfico
alt.Chart(geneTP53_counts.reset_index(name='counts')).mark_bar().encode(
    x='Grupo:N',
    y='counts:Q'
).properties(
    title='Contagem de TP53 por Grupo'
)
```

### 📝 **Problema 1.3**

**🎯 Objetivo:** Usando uma filtragem simples, faça um histograma comparando a quantidade de amostras de cada grupo de controle para o genes `BRCA1`.

```python
#TODO: Filtre o dados
geneBRCA1_df = df[df['Genes'] == 'BRCA1']

#TODO: Conte os dados de cada grupo
geneBRCA1_counts = geneBRCA1_df.groupby('Grupo').size()

#TODO: Crie o gráfico

#Dica: Siga os mesmos passos da problema anterior

alt.Chart(geneBRCA1_counts.reset_index(name='counts')).mark_bar().encode(
    x='Grupo:N',
    y='counts:Q'
).properties(
    title='Contagem de BRCA1 por Grupo'
)
```

### 📝 **Problema 1.4**
**🎯 Objetivo:** Visualize de forma organizada com o objetivo de verificar a relação do grupo de controle com a expressividade dos dois genes.

**OBS:** Este exercício admite muitas formas de criar, seja criativo.

```python
#TODO: Crie uma visualização

alt.Chart(df).mark_point().encode(
    y='Genes:N',
    x='Expressao:Q',
    color='Grupo:N',
    tooltip=['Amostra_ID', 'Genes', 'Expressao', 'Grupo']
).properties(
    title='Expressão de Genes por Grupo',
    height=150,
)
```

### 📝 **Problema 2.4**

**🎯 Objetivo:** Crie uma visualização bidimenional utilizando as colunas x e y para visualizar o trajeto do `Albert`.

**Nota:** Utiliza a biblioteca `altair` e seus métodos para criar a visualização.

```python
# TODO: faça uma visualização simples da trajetória do albert ao longo do tempo
alt.Chart(df).mark_line().encode(
    x='x:Q',
    y='y:Q',
).properties(
    title='Trajetória do Albert ao longo do tempo',
    width=500,
    height=300
)
```

### 📝 **Problema 2.5**

**🎯 Objetivo:** Crie uma coluna na estrutura de dados que mostra a distância percorrida entre as posições.

```python
# TODO: crie a coluna que motra a distância percorrida entre cada passo
df['distancia'] = np.sqrt((df['x'].diff() ** 2) + (df['y'].diff() ** 2))
df['distancia'] = df['distancia'].fillna(0)  # Preencher o primeiro valor com 0'

df.head()
```

### 📝 **Problema 2.6**

**🎯 Objetivo:** Crie uma visualização para mostrar a distância percorrida entre as posições ao longo do tempo.

```python
# TODO: faça uma visualização simples da trajetória do albert ao longo do tempo
alt.Chart(df).mark_line().encode(
    x='tempo:Q',
    y='distancia:Q',
).properties(
    title='Trajetória do Albert ao longo do tempo',
    width=500,
    height=300
)
```
