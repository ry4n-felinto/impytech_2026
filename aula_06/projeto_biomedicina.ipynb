Uma visualização interessante de se fazer para conhecermos os dados é o **boxplot**

```python
num_cols = df.select_dtypes(include='number').columns.tolist()
df_long = df.melt(value_vars=num_cols, var_name="Variavel", value_name="Valor")

boxplot_geral = alt.Chart(df_long).mark_boxplot().encode(
    y=alt.Y("Valor:Q", title=None),              # Passa os valores para o eixo Y
    color=alt.Color("Variavel:N", legend=None)
).properties(
    width=120,
    height=200
).facet(
    facet=alt.Facet("Variavel:N", title="Boxplot por Variável"),
    columns=4
).resolve_scale(
    y='independent'                             # Escala independente no eixo Y
)

boxplot_geral
```

## 5. Visualizações com Altair

Visualizações tem por objetivo **facilitar** o entendimento dos dados, se fazendo muito importante em contextos com maiores valores de icógnitas. Imagine por exemplo, que queremos ver a distribuição de frequência cardíaca máxima pela idade, agora temos diversas combinações de valor das variáveis. E se eu quisesse ver a relação desses 2 **preditores** com a presença de doença cardíaca? 

### 5.1 Idade x frequência cardíaca máxima, colorido pelo desfecho

Vamos fazer um scatter-plot utilizando as variáveis `Age` (Idade), `MaxHR` (Frequência cardíaca máxima) e `HeartDisease` (Problemas cardíacos). Utilizando as codifiações dos eixos X e Y para idade e frequência cardíaca máxima, respectivamente. Irei usar ambas as bibliotecas e fica a seu critério qual usar.

```python
# Usando matplotlib para criar um gráfico de dispersão (scatter plot) de Idade x MaxHR, colorido por HeartDisease
fig, ax = plt.subplots(figsize=(6, 4))

# Agrupando para criar a legenda separada por categoria de HeartDisease
for category, group in df.groupby("HeartDisease"):
    ax.scatter(
        group["Age"], 
        group["MaxHR"], 
        s=60, 
        alpha=0.5, 
        label=category
    )

# Títulos e rótulos
ax.set_title("Idade x MaxHR")
ax.set_xlabel("Idade")
ax.set_ylabel("Freq. cardíaca máxima")

# Legenda
ax.legend(title="Doença (1=sim)")

plt.tight_layout()
plt.show()
```

```python
# Por outro lado, podemos usar Altair para criar o mesmo gráfico de dispersão de forma interativa:
alt.Chart(df).mark_circle(size=60, opacity=0.5).encode(
    x=alt.X("Age:Q", title="Idade"),
    y=alt.Y("MaxHR:Q", title="Freq. cardíaca máxima"),
    color=alt.Color("HeartDisease:N", title="Doença (1=sim)"),
    tooltip=["Age", "MaxHR", "Sex", "ChestPainType"],
).interactive().properties(title="Idade x MaxHR", width=500)
```

É possível ver que não existe uma divisão clara de idade e frequência cardíaca máxima, visto que os dois grupos ainda estão entrelaçados entre si.
